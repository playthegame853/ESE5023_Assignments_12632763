```python
%matplotlib inline
import matplotlib.pyplot as plt

# ===================== 5.1 Find_expression =====================
def Find_expression(target):
    digits = "123456789"
    
    def dfs(pos, current_val, current_expr):
        if pos == len(digits):
            if current_val == target:
                print(f"{current_expr} = {target}")
            return
        num_str = ""
        for i in range(pos, len(digits)):
            num_str += digits[i]
            num = int(num_str)
            if pos == 0:
                dfs(i+1, num, num_str)
            else:
                dfs(i+1, current_val + num, current_expr + "+" + num_str)
                dfs(i+1, current_val - num, current_expr + "-" + num_str)
    dfs(0,0,"")


# ===================== 5.2 count solutions =====================
def count_solutions(target):
    digits = "123456789"
    count = 0
    def dfs(pos, current_val):
        nonlocal count
        if pos == len(digits):
            if current_val == target:
                count += 1
            return
        num_str = ""
        for i in range(pos, len(digits)):
            num_str += digits[i]
            num = int(num_str)
            if pos == 0:
                dfs(i+1, num)
            else:
                dfs(i+1, current_val + num)
                dfs(i+1, current_val - num)
    dfs(0,0)
    return count


# 生成 Total_solutions 列表，i从1~100
Total_solutions = []
for num in range(1,101):
    c = count_solutions(num)
    Total_solutions.append(c)
    print(f"i={num}, number of solutions = {c}")


# ====== 绘图 ======
plt.figure(figsize=(12,5))
plt.plot(range(1,101), Total_solutions)
plt.xlabel("i from 1 to 100")
plt.ylabel("Number of suitable solutions")
plt.title("Total_solutions for integers 1 ~ 100")
plt.grid(True)
plt.show()


# 找最大、最小值对应的数字
max_count = max(Total_solutions)
min_count = min(Total_solutions)
max_nums = [idx+1 for idx, val in enumerate(Total_solutions) if val == max_count]
min_nums = [idx+1 for idx, val in enumerate(Total_solutions) if val == min_count]

print(f"\nMaximum number of solutions = {max_count}, corresponding i: {max_nums}")
print(f"Minimum number of solutions = {min_count}, corresponding i: {min_nums}")


# 测试5.1样例
print("\n==== Find_expression(50) ====")
Find_expression(50)
```
